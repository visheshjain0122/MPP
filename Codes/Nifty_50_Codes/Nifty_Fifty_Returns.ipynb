```python
import pandas as pd
from nselib import capital_market
import requests
```

# Calling Data

```python
ret = pd.read_csv("/content/Nifty_Fifty_EOD_Data.csv")
ret = ret.set_index('date')
ret.head(10)
```

```python
#obtaining the nifty fifty list
data = capital_market.nifty50_equity_list()

#converting to a dataframe
df = pd.DataFrame(data)

#grabbing the stock symbols
symbols = df['Symbol'].tolist()

symbols
```

# Finding Returns

```python
#now to find returns using pct_change()

for i in range(1, len(symbols)+1):

  ret[symbols[i-1]] = ret[symbols[i-1]].pct_change()

  # 1. Drop NaN values
  ret = ret.dropna(subset=[symbols[i-1]])

  # 2. Convert to percentages
  ret[symbols[i-1]] = ret[symbols[i-1]] * 100

  # 3. Drop anything above 100%
  ret = ret[ret[symbols[i-1]] <= 100]
```

```python
ret
```

# Extracting as CSV

```python
csv_file_path = 'Nifty_Fifty_Returns.csv'  # Specify the path where you want to save the CSV file
ret.to_csv(csv_file_path)  # Set index=False to avoid writing row numbers

print(f'DataFrame has been saved to {csv_file_path}')
```
