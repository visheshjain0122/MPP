```python
import pandas as pd
from nselib import capital_market
import requests
```

# Extracting Nifty Fifty

```python
#obtaining the nifty fifty list
data = capital_market.nifty50_equity_list()

#converting to a dataframe
df = pd.DataFrame(data)

#grabbing the stock symbols
symbols = df['Symbol'].tolist()
```

```python
symbols
```

# Getting Closing Data

```python
# Initialize an empty list to store DataFrames for each symbol
dfs = {}

for i in range(1, len(symbols)+1):

    url = f'https://eodhd.com/api/eod/{symbols[i-1]}.NSE?from=2014-06-08&to=2024-06-08&api_token=5ffde981a31603.50022638&fmt=json' # with free API token, I can only do over the past year
    nifty_data = requests.get(url).json()

    #print(data)

    # Convert data to DataFrame
    dfs[f'df{i}'] = pd.DataFrame(nifty_data)

    # changing the date to the index column
    dfs[f'df{i}'] = dfs[f'df{i}'][['date', 'close']].set_index('date')

    # renaming the 'close' column to the name of the symbol
    dfs[f'df{i}'] = dfs[f'df{i}'].rename(columns={'close': symbols[i-1]})

    #print(dfs[f'df{i}'])



# Join the data frames in the dictionary
result_df = pd.DataFrame()

for df_name, df in dfs.items():
    result_df = result_df.join(df, how='outer')  # 'outer' joins all columns

print(result_df)
```

# Extracting as CSV

```python
csv_file_path = 'Nifty_Fifty_EOD_Data.csv'  # Specify the path where you want to save the CSV file
result_df.to_csv(csv_file_path)  # Set index=False to avoid writing row numbers

print(f'DataFrame has been saved to {csv_file_path}')
```
